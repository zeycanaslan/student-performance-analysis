```python
import pandas as pd

df= pd.read_csv('/kaggle/input/mall-customerscsv/Mall_Customers.csv')
df.head()
```

```python
import matplotlib.pyplot as plt

import pandas as pd
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.violinplot(df["Annual Income (k$)"])
ax.set_title("Annual Income Distribution - Violin Plot (Matplotlib OO)")
ax.set_ylabel("Annual Income (k$)")

plt.show()
```

```python
from matplotlib import pyplot as plt


plt.violinplot(df["Annual Income (k$)"])
plt.title("Annual Income Distribution - Violin Plot (Pyplot)")
plt.ylabel("Annual Income (k$)")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.violinplot(y=df["Annual Income (k$)"])
plt.title("Annual Income Distribution - Violin Plot (Seaborn)")
plt.ylabel("Annual Income (k$)")
plt.show()
```

```python
import plotly.express as px

fig = px.violin(
    df,
    y="Annual Income (k$)",
    box=True,
    points="all",
    title="Annual Income Distribution - Violin Plot (Plotly)"
)

fig.show()
```
