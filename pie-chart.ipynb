```python
import pandas as pd

df = pd.read_csv("/kaggle/input/world-population-dataset/world_population.csv")

# Kıtaya göre 2022 nüfusu toplamı
continent_pop = (
    df.groupby("Continent")["2022 Population"]
    .sum()
    .reset_index()
)

continent_pop
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.pie(
    continent_pop["2022 Population"],
    labels=continent_pop["Continent"],
    autopct="%1.1f%%",
    startangle=90
)

ax.set_title("World Population by Continent (2022) - Matplotlib OO")
plt.show()

```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

plt.pie(
    continent_pop["2022 Population"],
    labels=continent_pop["Continent"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("World Population by Continent (2022) - Seaborn Style")
plt.show()
```
