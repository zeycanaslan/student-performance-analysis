```python
import pandas as pd
import matplotlib.pyplot as plt

# CSV dosyasını oku
df = pd.read_csv(
    "/kaggle/input/students-performance-in-exams/StudentsPerformance.csv"
)

# Matematik notlar
math_scores = df["math score"]
```

```python

plt.figure()
plt.boxplot(math_scores)
plt.title("Matematik Notları - Box Plot (Matplotlib, pyplot)")
plt.ylabel("Not Değeri")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure()
sns.boxplot(y=math_scores)
plt.title("Matematik Notları - Box Plot (Seaborn)")
plt.ylabel("Not Değeri")
plt.show()
```

```python
import plotly.express as px

fig = px.box(
    y=math_scores,
    title="Matematik Notları - Box Plot (Plotly)",
    labels={"y": "Not Değeri"}
)
fig.show()
```
